Sum of numbers in list

```python
def add_many_numbers(numbers)-> int:
  total = 0
  for i in numbers:
      total += i
  return total

def main():
    numbers: list[int] = [1, 2, 3, 4, 5]
    print(add_many_numbers(numbers))


if __name__ == '__main__':
    main()
```

03_erase_canvas

```python
import tkinter as tk

CANVAS_WIDTH = 400
CANVAS_HEIGHT = 400
CELL_SIZE = 40
ERASER_SIZE = 20

class EraserApp:
    def __init__(self, root):
        self.canvas = tk.Canvas(root, width=CANVAS_WIDTH, height=CANVAS_HEIGHT, bg='white')
        self.canvas.pack()
        self.cells = []

        # Draw the grid of blue cells
        for row in range(0, CANVAS_HEIGHT, CELL_SIZE):
            for col in range(0, CANVAS_WIDTH, CELL_SIZE):
                cell = self.canvas.create_rectangle(
                    col, row,
                    col + CELL_SIZE, row + CELL_SIZE,
                    fill='blue', outline='black'
                )
                self.cells.append(cell)

        # Create the eraser rectangle
        self.eraser = self.canvas.create_rectangle(0, 0, ERASER_SIZE, ERASER_SIZE, fill='pink')

        # Bind mouse movement
        self.canvas.bind("<B1-Motion>", self.move_eraser)

    def move_eraser(self, event):
        # Move eraser to follow mouse
        x1 = event.x
        y1 = event.y
        x2 = x1 + ERASER_SIZE
        y2 = y1 + ERASER_SIZE
        self.canvas.coords(self.eraser, x1, y1, x2, y2)

        # Check for overlapping items
        overlapping = self.canvas.find_overlapping(x1, y1, x2, y2)
        for obj in overlapping:
            if obj != self.eraser:
                self.canvas.itemconfig(obj, fill='white')

def main():
    root = tk.Tk()
    root.title("Canvas Eraser")
    app = EraserApp(root)
    root.mainloop()

if __name__ == "__main__":
    main()
```

04_flowing_with_data_structures

```python
def copies(my_list, data):
    for _ in range(3):
        my_list.append(data)

def main():
    message = input("Enter to copy: ")

    lst = []
    print("List before:", lst)

    copies(lst, message)

    print("List after:", lst)

if __name__ == '__main__':
    main()
```

```python
def main():
    lst = []

    val = input("Enter a value: ")
    while val != "":
        lst.append(val)
        val = input("Enter a value: ")

    print("Here's the list:", lst)

if __name__ == '__main__':
    main()
```

shorten lst

```python
def shorten(lst):
    while len(lst) > 3:
        removed = lst.pop()
        print(removed)

def main():
    my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']
    print("Original list:", my_list)

    shorten(my_list)

    print("Shortened list:", my_list)

if __name__ == '__main__':
    main()
```
