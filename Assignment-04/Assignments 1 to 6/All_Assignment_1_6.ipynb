```python
import random
# guess the number

def guess(x):
    random_number = random.randint(1, x)
    guess = 0
    while guess != random_number:
        guess = int(input(f"Guess a number between 1 and {x}: "))
        if guess < random_number:
            print("Sorry, guess again. Too low.")
        elif guess > random_number:
            print("Sorry, guess again. Too high.")
    print(f"Yay, congrats. You have guessed the number {random_number} correctly!")

guess(20)
```

```python
import random

def guess(x):
    low = 1
    high = x
    feedback = ''
    guess = int(input(f"take a number between {low} and {high}: "))
    while feedback != 'c':
        random_number = random.randint(low, high)
        if guess < random_number:
            print("Sorry, guess again. Too high.")
            high = high - 1
        elif guess > random_number:
            print("Sorry, guess again. Too low.")
            low = low + 1
        else:
            print(f"Yay, congrats. You have guessed the number {random_number} correctly!")
            feedback = 'c'

guess(20)
```

```python
import string
import random

words = ['hello', "world", "yoyo", "pakistan", "computer", "lol", "ASAP", "hi"]

def hangman():
    word_choice = random.choice(words).upper()
    word_letters = set(word_choice)
    used_letters = set()
    alphabet = set(string.ascii_uppercase)

    print("Welcome to Hangman!")

    while len(word_letters) > 0:
        print("\nYou have Used letters:", " ".join(sorted(used_letters)))

        word_display = [letter if letter in used_letters else '_' for letter in word_choice]
        print("Current word: ", " ".join(word_display))

        user_in = input("Guess a letter: ").upper()

        if user_in in alphabet - used_letters:
            used_letters.add(user_in)
            if user_in in word_letters:
                word_letters.remove(user_in)
                print("Correct!")
            else:
                print("Incorrect!")
        elif user_in in used_letters:
            print("You already guessed that letter.")
        else:
            print("Invalid input. Please enter a letter.")

hangman()
```
