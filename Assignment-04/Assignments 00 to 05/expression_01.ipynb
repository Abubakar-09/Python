```python
import random

def main():
    dice_roll()
    dice_roll()
    dice_roll()

def dice_roll():
    dice1 = random.randint(1,6)
    dice2 = random.randint(1,6)
    print("total of two dice =",dice1+dice2)

# This provided line is required at the end of
# Python file to call the main() function.
if __name__ == '__main__':
    main()
```

03_feet_to_inches

```python
INCHES_IN_FOOT: int = 12

def main():
    feet: float = float(input("Enter number of feet: "))
    inches: float = feet * INCHES_IN_FOOT
    print("That is", inches, "inches!")

if __name__ == '__main__':
    main()
```

04_pythagorean_theorem

```python
import math

def main():
    ab: float = float(input("Enter the length of Base: "))
    ac: float = float(input("Enter the length of Perpendicular: "))
    bc: float = math.sqrt(ab**2 + ac**2)
    print("The length of Hypotenuse is", bc)

if __name__ == '__main__':
    main()
```

05_remainder_division

```python
def main():
    first: int = int(input("Please enter an integer to be divided: "))
    second: int = int(input("Please enter an integer to divide by: "))

    division_op: int = first // second
    remainder_op: int = first % second

    print(f"{first} divided by {second} is {division_op} with remainder {remainder_op}")


if __name__ == '__main__':
    main()
```

06_rolldice

```python
def main():
    roll_dice()

def roll_dice():
    import random
    dice1 = random.randint(1,6)
    dice2 = random.randint(1,6)

    print(f"Dice 1 is = {dice1}")
    print(f"Dice 2 is = {dice2}")
    print(f"total is = {dice1+dice2}")

if __name__ == '__main__':
    main()
```

06_seconds_in_year

```python
DAYS_PER_YEAR: int = 365
HOURS_PER_DAY: int = 24
MIN_PER_HOUR: int = 60
SEC_PER_MIN: int = 60

def main():
    print(f"Number of seconds in a year is {DAYS_PER_YEAR*HOURS_PER_DAY*MIN_PER_HOUR*SEC_PER_MIN}")

if __name__ == '__main__':
    main()
```
