```python
def main():
    earth_weight = float(input("Enter a weight on Earth: "))

    planet = input("Enter a planet: ").lower()

    gravity = {
        "mercury": 0.376,
        "venus": 0.889,
        "mars": 0.378,
        "jupiter": 2.36,
        "saturn": 1.081,
        "uranus": 0.815,
        "neptune": 1.14
    }

    if planet in gravity:
        planet_weight = round(earth_weight * gravity[planet], 2)
        print(f"The equivalent weight on {planet}: {planet_weight}")

if __name__ == "__main__":
    main()
```

```python
def get_lst_element(lst, index):
    try:
        return lst[index]
    except IndexError:
        return "Index out of range."

def modify_lst_element(lst, index, new_value):
    try:
        lst[index] = new_value
        return lst
    except IndexError:
        return "Index out of range."

def slice_list(lst, start, end):
    try:
        return lst[start:end]
    except:
        return "Invalid slicing indices."

def game():
    lst = ['red', 'blue', 'green', 'yellow', 'purple']
    while True:
        print("\nCurrent list:",lst)
        print("Choose an operation:")
        print("1. Access an element")
        print("2. Modify an element")
        print("3. Slice the list")
        print("4. Quit")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            idx = int(input("Enter index to access: "))
            print("Result:", get_lst_element(lst, idx))

        elif choice == "2":
            idx = int(input("Enter index to modify: "))
            new_val = input("Enter new value: ")
            result = modify_lst_element(lst, idx, new_val)
            print("Result:", result)

        elif choice == "3":
            start = int(input("Enter start index: "))
            end = int(input("Enter end index: "))
            print("Sliced List:", slice_list(lst, start, end))

        elif choice == "4":
            print("Goodbye!")
            break

        else:
            print("Invalid choice. Try again.")

if __name__ == "__main__":
    game()
```
